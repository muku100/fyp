```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/bengaluru_house_prices.csv")
print("shape:", df.shape)
print(df.columns.tolist())
df.head()
```

```python
print(df.isnull().sum())
print("society unique:", df["society"].nunique())
print("society missing %:", round(df["society"].isna().mean()*100,1))
print("location unique:", df["location"].nunique())
print(df["size"].value_counts(dropna=False).head(20))
```

```python
mask = df["total_sqft"].astype(str).str.contains("-", na=False)
print("range rows:", mask.sum())
print(df.loc[mask, "total_sqft"].head(10).tolist())
print("dups:", df.duplicated().sum())
```
