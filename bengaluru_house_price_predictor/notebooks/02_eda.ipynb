# 02 - EDA (Exploratory Data Analysis)
Goal: see distributions, outliers, location/society patterns BEFORE modeling.
`price_per_sqft` here is EDA-ONLY (it uses price, so never put it in X).

```python
import sys, os
sys.path.append(os.path.abspath('..'))
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('../data/raw/bengaluru_house_prices.csv')
print(df.shape)
print(df['price'].describe())
plt.figure(figsize=(10,4))
plt.hist(df['price'].clip(upper=500), bins=50)
plt.xlabel('Price (lakhs, clipped at 500)')
plt.ylabel('Count')
plt.title('Price distribution (clipped - outliers exist above 500)')
plt.show()
```

```python
# sqft vs price (sample 2000 pts so plot is fast)
import numpy as np
s = df.sample(2000, random_state=42)
plt.figure(figsize=(8,5))
plt.scatter(s['bath'], s['price'], alpha=0.3)
plt.xlabel('bath')
plt.ylabel('price (lakhs)')
plt.title('bath vs price')
plt.show()
print('price>1000:', (df['price']>1000).sum())
print('bath>10:', (df['bath']>10).sum())
print('balcony value counts:')
print(df['balcony'].value_counts(dropna=False).head(10))
```

```python
# location: counts vs mean price
counts = df['location'].value_counts().head(20)
print(counts)
mean_price = df.groupby('location')['price'].mean().sort_values(ascending=False).head(10)
print('\nTop 10 priciest locations (mean price):')
print(mean_price.round(1))
```

```python
from src.data_cleaning import convert_sqft
print('society unique:', df['society'].nunique())
print('society missing %:', round(df['society'].isna().mean()*100,1))
print(df['society'].value_counts().head(10))
# price_per_sqft EDA ONLY (uses proper convert_sqft; 46 rows with Sq.Meter/Acres units -> NaN, dropped later)
sqft_num = df['total_sqft'].apply(convert_sqft)
print('unparseable sqft (units like Sq. Meter/Acres):', int(sqft_num.isna().sum()))
pp = df['price']*100000/sqft_num
print('rough price_per_sqft describe (EDA only, do NOT use as feature):')
print(pp.describe())
print('locations with <10 samples:', (df['location'].value_counts()<10).sum(), '-> rare-grouping needed in OneHot')
```
